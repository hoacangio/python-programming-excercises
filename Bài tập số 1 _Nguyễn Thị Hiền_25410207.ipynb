### 2. Viết chương trình nhập vào 2 số nguyên a, b. Tính tổng, hiệu, tích,
thương, chia lấy dư, chia lấy nguyên của 2 số trên và in kết quả ra màn
hình. Kết quả phép chia làm tròn 2, 3 chữ số sau dấu chấm (ví dụ kết
quả 3.333333 thì làm tròn 3.333).

```python
a = int(input("nhap so a: "));
b = int(input("nhap so b: "));

tong = a + b;
hieu = a - b;
tich = a * b;

print("Tong cua 2 so la: ", tong);
print("Hieu cua 2 so la: ", hieu);
print("Tich cua 2 so la: ", tich);

if b != 0:
    thuong = round(a / b,3);
    chiaLayDu = a % b;
    chiaLayNguyen = a // b;

    print("Thuong cua 2 so la: ", thuong);
    print("Chia Lay phan du cua 2 so la: ", chiaLayDu);
    print("Chia Lay phan nguyen cua 2 so la: ", chiaLayNguyen);
else :
    print("Khong the chia cho 0");
```

### 7. Viết chương trình in ra menu

```python
print("============ MENU ============")
print("1. Hu tieu");
print("2. Chao long");
print("3. Banh canh");
print("4. Bun rieu");
print("5. Pho bo");
print("==============================")

# Switch - case
luachon = int(input("nhap luachon: "));

match luachon:
    case 1:
        print("Hu tieu");
    case 2:
        print("Chao long");
    case 3:
        print("Banh canh");
    case 4:
        print("Bun rieu");
    case 5:
            print("Pho bo");
    case _:
            print("Lua chon khong hop le");
print("==============================")
```

### 8. Nhập bán kính của đường tròn, tính và in ra chu vi, diện tích của hình tròn.

```python
import math;

banKinh = float(input("nhap ban kinh: "));

chuVi = 2 * math.pi * banKinh;
dienTich =  math.pi * banKinh ** 2;

print("Chu vi hinh tron la:", round(chuVi, 2))
print("Dien tich hinh tron la:", round(dienTich, 2))
```

### 9. Viết chương trình tính số đo kiểm tra sức khỏe BMI.
Công thức BMI = 𝐶â𝑛_𝑛ặ𝑛𝑔 (𝑘𝑔)/𝐶ℎ𝑖ề𝑢_𝑐𝑎𝑜2 (𝑚)

```python
kg = float(input("Nhap can nang (kg): "));
m = float(input("Nhap chieu cao (m): "));

BMI = kg / (m ** 2)

print("Chi so BMI cua ban la:", round(BMI, 2))
```

### 10. Nhập vào số xe(gồm 5 chữ số). Cho biết số xe có mấy nút

```python
soXe = int(input("Nhap so xe (5 chu so): "))
print("So xe la: ", soXe)

sum = 0;
while soXe > 0:
    sum += soXe % 10; # tinh tong so
    soXe = soXe // 10; # lay phan nguyen bo phan du da tinh
nut = sum % 10; # so nut o phan du

print("Xe co so nut la: ", nut);
```

### 12.Viết chương trình xuất ra số (nguyên, thực) ngẫu nhiên theo yêu cầu sau:
• 0 đến100
• 50 đến99
•-39 đến 79
•-79 đến-39

```python
import random;

soNgauNhien = random.randint(0,100);
soNgauNhien1 = random.randint(50,99);
soNguyenAm = random.randint(-39,79);
soNguyenAm2 = random.randint(-79,-39);

print("So ngau nhien tu 0 den 100:", round(soNgauNhien,2));
print("So ngau nhien tu 50 den 99:", round(soNgauNhien1,2));
print("So ngau nhien tu -39 den 79:", round(soNguyenAm,2));
print("So ngau nhien tu -79 den -39:", round(soNguyenAm2,2));
```
