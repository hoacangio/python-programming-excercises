```python
# BÀI TẬP TỔNG HỢP - BÀI #3
# Kiểm tra User ID và Password


# Hàm kiểm tra User ID
def kiem_tra_user_id(user_id):
    # Kiểm tra độ dài
    if len(user_id) < 6 or len(user_id) > 24:
        print("User ID phải có từ 6 đến 24 ký tự.")
        return False

    # Kiểm tra khoảng trắng
    if " " in user_id:
        print("User ID không được chứa khoảng trắng.")
        return False

    # Các ký tự không được phép
    ky_tu_cam = "!@#$%^&*()-=+"

    for ky_tu in user_id:
        if ky_tu in ky_tu_cam:
            print("User ID không được chứa ký tự đặc biệt:", ky_tu)
            return False

    return True


# Hàm kiểm tra Password
def kiem_tra_password(password):
    # Kiểm tra độ dài
    if len(password) < 6 or len(password) > 24:
        print("Password phải có từ 6 đến 24 ký tự.")
        return False

    # Kiểm tra ít nhất 1 chữ thường
    co_chu_thuong = any(ky_tu.islower() for ky_tu in password)

    # Kiểm tra ít nhất 1 chữ hoa
    co_chu_hoa = any(ky_tu.isupper() for ky_tu in password)

    # Kiểm tra ít nhất 1 chữ số
    co_chu_so = any(ky_tu.isdigit() for ky_tu in password)

    # Kiểm tra ít nhất 1 ký tự đặc biệt $, #, @
    co_ky_tu_dac_biet = any(ky_tu in "$#@" for ky_tu in password)

    if not co_chu_thuong:
        print("Password phải có ít nhất 1 chữ cái thường [a-z].")
        return False

    if not co_chu_so:
        print("Password phải có ít nhất 1 chữ số [0-9].")
        return False

    if not co_chu_hoa:
        print("Password phải có ít nhất 1 chữ cái hoa [A-Z].")
        return False

    if not co_ky_tu_dac_biet:
        print("Password phải có ít nhất 1 ký tự đặc biệt [$ # @].")
        return False

    return True


# Chương trình chính
user_id = input("Nhập User ID: ")
password = input("Nhập Password: ")

if kiem_tra_user_id(user_id) and kiem_tra_password(password):
    print("Đăng nhập hợp lệ!")
else:
    print("Đăng nhập không hợp lệ!")
```
